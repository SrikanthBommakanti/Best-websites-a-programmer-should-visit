```python
import itertools

def compress_the_string(string):
    # return empty string when input is empty
    if not string:
        return ""
    parts = []
    for k, g in itertools.groupby(string):
        parts.append(f"({len(list(g)), k})")
    return " ".join(parts)

if __name__ == '__main__':
    string = "mississippi"
    print(compress_the_string(string))
```

```python
import itertools

def compress_groupby(string):
    if not string:
        return ""
    parts = []
    for ch, group in itertools.groupby(string):
        count = sum(1 for _ in group)   # no list allocation
        parts.append(ch if count == 1 else f"{ch}{count}")
    return "".join(parts)

if __name__ == '__main__':
    print(compress_groupby("mississippi"))  # -> mis2is2ip2i
```

```python
def compress_manual(s):
    if not s:
        return ""
    parts = []
    prev = s[0]
    count = 1
    for ch in s[1:]:
        if ch == prev:
            count += 1
        else:
            parts.append(prev if count == 1 else f"{prev}{count}")
            prev = ch
            count = 1
    parts.append(prev if count == 1 else f"{prev}{count}")  # last group
    return "".join(parts)

if __name__ == '__main__':
    print(compress_manual("mississippi"))  # -> mis2is2ip2i
```

```python
import timeit

test_string = "mississippi" * 1000 # Use a longer string for better benchmarking

# Benchmark compress_the_string
time_compress_the_string = timeit.timeit(
    "compress_the_string(test_string)",
    globals=globals(),
    number=1000
)
print(f"Time taken by compress_the_string: {time_compress_the_string:.6f} seconds")

# Benchmark compress_groupby
time_compress_groupby = timeit.timeit(
    "compress_groupby(test_string)",
    globals=globals(),
    number=1000
)
print(f"Time taken by compress_groupby: {time_compress_groupby:.6f} seconds")

# Benchmark compress_manual
time_compress_manual = timeit.timeit(
    "compress_manual(test_string)",
    globals=globals(),
    number=1000
)
print(f"Time taken by compress_manual: {time_compress_manual:.6f} seconds")
```
